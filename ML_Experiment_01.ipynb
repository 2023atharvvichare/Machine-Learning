```python
import numpy as np
import pandas as pd
```

```python
names = pd.read_csv("Indian_Names.csv")
```

```python
df = pd.DataFrame(names)
```

```python
df.dropna(inplace=True)
```

```python
df = df['Name'].str.replace('[^a-zA-Z]', '' , regex=True)
df[:5]
```

```python
df.loc[0:6]
```

```python
df.info()
```

```python
names_ = list(df.dropna().astype(str).str.split(' '))
```

```python
names_[:5]
```

```python
names_ = [name for sublist in names_ for name in sublist]
np_array = np.array(names_)
print(f"Original NumPy array shape: {np_array.shape}")
names_1D = np_array.flatten()
print(f"Flattened NumPy array (using flatten()): {names_1D}")
print(f"Flattened NumPy array shape (using flatten()): {names_1D.shape}")
```

```python
aplha = sorted(list(set(''.join(names_1D))))
stoi = {s:i+1 for i,s in enumerate(aplha)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
itos
```

```python
X = []
Y = []
content = [0]*3
for word in names_1D:
  word = '.' + word + '.'
  for chr in word:
      ix = stoi[chr]
      X.append(content)
      Y.append(ix)
      content = content[1:] + [ix]
```

```python
X[:5] , Y[:5]
```

```python
X = np.array(X)
Y = np.array(Y)
```

```python
Y.shape
```

```python
w1 = np.random.rand(3, 400)*(3/5)*3
```

```python
w1
```

```python
logits = np.dot(X, w1)
logits
```

```python
layer2 = np.tanh(logits)
```

```python
layer2
```

```python
res = (layer2 - layer2.min()) / (layer2.max() - layer2.min())
```
